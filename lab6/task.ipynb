### Решение Задания 1: Алгоритм Дойча


```python
# Решение Задания 1 (Алгоритм Дойча)
from qiskit import QuantumCircuit
import numpy as np

def twobit_function(oracle_type: int):
    # Создание квантовой схемы оракула U_f на 2 кубитах (q0 - вход, q1 - выход)
    qc = QuantumCircuit(2, name=f'U_f{oracle_type}')
    if oracle_type == 1:
        pass # f(x) = 0
    elif oracle_type == 2:
        qc.cx(0, 1) # f(x) = x
    elif oracle_type == 3:
        qc.x(0)
        qc.cx(0, 1)
        qc.x(0) # f(x) = NOT x
    elif oracle_type == 4:
        qc.x(1) # f(x) = 1
    return qc

def deutsch_decide(oracle_qc):
    qc = QuantumCircuit(2, 1)
    # Подготовка кубитов: q0 в |+>, q1 в |->
    qc.x(1)
    qc.h(0)
    qc.h(1)
    # Применение оракула как чёрного ящика ровно 1 раз
    qc.append(oracle_qc.to_gate(), [0, 1])
    # Финальный H на управляющем кубите и измерение
    qc.h(0)
    qc.measure(0, 0)
    return qc

# Проверка на всех 4 функциях
print('Оракул | Тип функции (ожид) | Статус проверки')
print('-------+--------------------+----------------')
expected_types = {1: 'постоянная', 2: 'сбалансированная', 3: 'сбалансированная', 4: 'постоянная'}
for t in (1, 2, 3, 4):
    oracle = twobit_function(t)
    circ = deutsch_decide(oracle)
    assert circ.data[3].operation.name == f'U_f{t}' # Ровно один вызов оракула
    print(f'  f_{t}   | {expected_types[t]:<18} | OK (вызван ровно 1 раз)')
print('[+] Все оракулы проверены assert!')
```

### Ответы на теоретические вопросы Задания 1:
1. **Почему результат «1» на первом кубите соответствует сбалансированной функции?**
   Благодаря фазовому сдвигу (phase kickback) от целевого кубита в состоянии $|-angle$, оракул переводит $|xangle$ в -1^{f(x)}|xangle$. После оракула состояние управляющего кубита равно $\frac{(-1)^{f(0)}|0\rangle + (-1)^{f(1)}|1\rangle}{\sqrt{2}}$. Если функция сбалансирована ((0) \neq f(1)$), состояние имеет вид $\pm |-\rangle$, и финальный вентиль Адамара переводит его строго в состояние $|1\rangle$.
2. **Как оракул, не меняя значений кубита 0 в вычислительном базисе, меняет его состояние?**
   Оракул действует как |x\rangle|y\rangle = |x\rangle|y \oplus f(x)\rangle$. Когда целевой кубит подготовлен в суперпозиции $|-\rangle = \frac{|0\rangle - |1\rangle}{\sqrt{2}}$, действие  \oplus f(x)$ умножает базисные векторы на фазовый множитель -1^{f(x)}$. Эта относительная фаза между компонентами суперпозиции кубита 0 изменяет интерференцию при последующем применении вентиля Адамара, хотя в вычислительном базисе значение $ не изменялось.


### Решение Задания 2: Алгоритм Дойча-Йожи


```python
# Решение Задания 2: Алгоритм Дойча-Йожи
def dj_sim(n, f_type='const'):
    # f_type: 'const' или 'balanced'
    N = 1 << n
    if f_type == 'const':
        f = lambda x: 1
    else:
        f = lambda x: x & 1
    amp_zeros = sum((-1)**f(x) for x in range(N)) / N
    prob_zeros = amp_zeros ** 2
    return prob_zeros

print('n | P(|0...0>) для const | P(|0...0>) для balanced | Решение алгоритма')
print('--+----------------------+-------------------------+-------------------')
for n in (2, 3, 4):
    p_c = dj_sim(n, 'const')
    p_b = dj_sim(n, 'balanced')
    assert np.isclose(p_c, 1.0)
    assert np.isclose(p_b, 0.0)
    print(f'{n} |        {p_c:.4f}        |         {p_b:.4f}          | Успешно разделено')
print('[+] Алгоритм Дойча-Йожи гарантирует 100% определение типа функции за 1 квантовое обращение!')
```

### Решение Заданий 3-6: Алгоритм Шора (Вариант 7: Таблица А, N = 35, a = 2)


```python
# Решение Заданий 3-6 (Вариант 7)
import math
from fractions import Fraction

N = 35
a = 2

# 1. Проверка gcd(a, N) = 1
g = math.gcd(a, N)
assert g == 1, f'Числа не взаимно просты! gcd = {g}'
print(f'1. gcd({a}, {N}) = {g} (взаимно просты)')

# 2. Поиск периода r
r = 1
while pow(a, r, N) != 1:
    r += 1
assert r == 12
print(f'2. Порядок числа a={a} по модулю {N}: r = {r}')

# 3. Квантовые параметры схемы
n = math.ceil(math.log2(N)) # 6 кубитов
m = 2 * n                    # 12 управляющих кубитов
two_m = 1 << m               # 4096
print(f'3. Параметры: регистр фазы m={m} (2^m={two_m}), целевой регистр n={n}')

# 4. Анализ квантовых измерений
def euler_phi(x):
    res = x
    p = 2
    temp = x
    while p * p <= temp:
        if temp % p == 0:
            while temp % p == 0:
                temp //= p
            res -= res // p
        p += 1
    if temp > 1:
        res -= res // temp
    return res

print('\nТаблица восстановления порядка r из фазовой оценки y/2^m:')
print(' s  | Фаза s/r | Дискретный y | Дробь y/2^m | Кандидат r | Статус')
print('----+----------+--------------+-------------+------------+------------------')

successes = 0
for s in range(r):
    phase = s / r
    y = round(phase * two_m)
    frac = Fraction(y, two_m).limit_denominator(N)
    cand_r = frac.denominator
    is_valid = (cand_r > 1 and pow(a, cand_r, N) == 1)
    if is_valid:
        successes += 1
        status = f'Засчитано (r={cand_r})'
    else:
        status = 'Делитель r (или s=0)'
    print(f'{s:2d}  |  {phase:.4f}  |     {y:4d}     |   {str(frac):>7}   |     {cand_r:2d}     | {status}')

rate = successes / r
theor_rate = euler_phi(r) / r
print(f'\nДоля успешных измерений: {rate * 100:.2f}%')
print(f'Теоретическая величина phi(r)/r = {euler_phi(r)}/{r} = {theor_rate * 100:.2f}%')
assert math.isclose(rate, theor_rate)

# 5. Факторизация числа N = 35
assert r % 2 == 0
half_r = r // 2
x = pow(a, half_r, N) # 2^6 mod 35 = 64 mod 35 = 29
print(f'\n5. x = a^(r/2) mod N = 2^{half_r} mod {N} = {x}')
assert x != N - 1 and x != 1

d1 = math.gcd(x - 1, N)
d2 = math.gcd(x + 1, N)
print(f'   d1 = gcd(x - 1, N) = gcd({x - 1}, {N}) = {d1}')
print(f'   d2 = gcd(x + 1, N) = gcd({x + 1}, {N}) = {d2}')
assert d1 * d2 == N
assert 1 < d1 < N and 1 < d2 < N
print(f'\n[+] УСПЕХ: Найдены нетривиальные делители числа {N}: {d1} и {d2}!')
```

1. **Что произойдёт, если взять простое N, например N = 1009? Может ли алгоритм дать нетривиальный множитель, и почему?**
   Алгоритм не даст нетривиальных множителей, так как у простого числа $ по определению нет делителей, кроме 1 и самого $. Вычисление $\gcd(a^{r/2} \pm 1, N)$ даст либо 1, либо $.
2. **Почему основание с нечётным порядком r не даёт множителей?**
   Алгоритм Шора основан на разложении разности квадратов: ^r - 1 = (a^{r/2} - 1)(a^{r/2} + 1) = k \cdot N$. Если порядок $ нечётен, то /2$ не является целым числом, и такое алгебраическое разложение на целые сомножители в кольце вычетов невозможно.
